Colocar estatus genericos a status y añadir columna de estatus generico a results

```python
ids_finished = [1, 11, 12, 13, 14, 111, 128, 88, 19, 15, 45, 50, 55, 17, 58, 16, 18]
ids_mechanical = [26, 5, 10, 23, 25, 75, 44, 38, 39, 6, 65, 95, 21,
                    7, 8, 9, 22, 105, 141, 103, 27, 28, 101, 135, 30,
                    132, 131, 32, 34, 98, 99, 93, 94, 129, 76, 36, 37, 40,
                    42, 43, 51, 56, 66, 84, 67, 85, 47, 48, 79, 86, 72, 69, 74,
                    61, 70, 87, 63, 80, 91, 24, 46, 60, 71]
ids_accident = [130, 3, 4, 20, 137, 138, 49, 83, 140, 33, 41, 136, 29, 59]
ids_retired = [31]
ids_other = [54, 139, 2, 96, 97, 62, 77, 81, 90, 92, 89, 73, 82, 68, 78, 35, 64] 

condiciones = [
    status['statusId'].isin(ids_finished),
    status['statusId'].isin(ids_mechanical),
    status['statusId'].isin(ids_accident),
    status['statusId'].isin(ids_retired),
    status['statusId'].isin(ids_other)
]

categorias = ['Finished', 'Mechanical issue', 'Accident', 'Retired', 'Other']

status['statusGeneric'] = np.select(condiciones, categorias, default=None)
status
```

```python
results_23 = results_23.drop(columns=['status'], errors='ignore')
results_23 = results_23.merge(
    status[['statusId', 'statusGeneric']], 
    on='statusId', 
    how='left'
)
```
